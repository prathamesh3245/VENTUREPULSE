```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel('Final_Financial_Dataset_Practical.xlsx')
df.head()
```

```python
monthly_revenue = []

monthly_revenue.append(df[df['name'] == '#waywire']['month_1_revenue'].values[0])
monthly_revenue.append(df[df['name'] == '#waywire']['month_2_revenue'].values[0])
monthly_revenue.append(df[df['name'] == '#waywire']['month_3_revenue'].values[0])
monthly_revenue.append(df[df['name'] == '#waywire']['month_4_revenue'].values[0])
monthly_revenue.append(df[df['name'] == '#waywire']['month_5_revenue'].values[0])
monthly_revenue.append(df[df['name'] == '#waywire']['month_6_revenue'].values[0])
```

```python
monthly_revenue
```

```python
print(monthly_revenue)
```

```python

x = np.array(monthly_revenue)
y = np.array([1, 2, 3, 4, 5, 6]) 
```

```python
plt.plot(y , x)
plt.xlabel('Months')
plt.ylabel('Revenue in $')
plt.title('Monthly Revenue for #waywire')
plt.show()
```

```python
plt.fill_between(y , x)
plt.xlabel('Months')
plt.ylabel('Revenue in $')
plt.title('Monthly Revenue for #waywire')
plt.show()
```
