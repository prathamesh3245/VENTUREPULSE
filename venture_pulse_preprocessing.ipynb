```python
import numpy as np
import pandas as pd

# ------------------------------
# STEP 1: Load dataset
# ------------------------------
df = pd.read_excel("Final_Financial_Dataset_Practical.xlsx")

# ------------------------------
# STEP 2: Compute Net Revenue
# ------------------------------
# Formula: Net Revenue = sum of month_i_revenue - Discounts - Returns & Allowances
df['Net Revenue'] = (
    df[['month_1_revenue','month_2_revenue','month_3_revenue',
        'month_4_revenue','month_5_revenue','month_6_revenue']].sum(axis=1)
    - df['Discounts']
    - df['Returns & Allowances']
)

# ------------------------------
# STEP 3: Select relevant features
# ------------------------------
features = [
    'Operating Expense', 'COGS', 'Discounts', 'Units Sold',
    'MAU/AAU', 'Churn Rate', 'Expansion Revenue',
    'month_1_revenue', 'month_2_revenue', 'month_3_revenue',
    'month_4_revenue', 'month_5_revenue', 'month_6_revenue'
]
target = 'Net Revenue'

# Drop missing values
df = df.dropna(subset=features + [target])

X = df[features].values
y = df[target].values

# ------------------------------
# STEP 4: Implement simple XGBoost-like model from scratch
# ------------------------------
class SimpleXGBoostRegressor:
    def __init__(self, n_estimators=10, learning_rate=0.1, max_depth=3):
        self.n_estimators = n_estimators
        self.learning_rate = learning_rate
        self.max_depth = max_depth
        self.trees = []
        self.base_pred = None

    def _split(self, X, y, depth):
        if depth == self.max_depth or len(np.unique(y)) == 1:
            return np.mean(y)

        best_feature, best_thresh, best_loss = None, None, float('inf')
        n_samples, n_features = X.shape

        for feature in range(n_features):
            thresholds = np.unique(X[:, feature])
            for thresh in thresholds:
                left_mask = X[:, feature] <= thresh
                right_mask = X[:, feature] > thresh
                if len(y[left_mask]) == 0 or len(y[right_mask]) == 0:
                    continue
                loss = np.var(y[left_mask]) * len(y[left_mask]) + np.var(y[right_mask]) * len(y[right_mask])
                if loss < best_loss:
                    best_feature, best_thresh, best_loss = feature, thresh, loss

        left_mask = X[:, best_feature] <= best_thresh
        right_mask = X[:, best_feature] > best_thresh

        return {
            'feature': best_feature,
            'threshold': best_thresh,
            'left': self._split(X[left_mask], y[left_mask], depth + 1),
            'right': self._split(X[right_mask], y[right_mask], depth + 1)
        }

    def _predict_tree(self, tree, x):
        if not isinstance(tree, dict):
            return tree
        if x[tree['feature']] <= tree['threshold']:
            return self._predict_tree(tree['left'], x)
        else:
            return self._predict_tree(tree['right'], x)

    def fit(self, X, y):
        self.base_pred = np.mean(y)
        y_pred = np.full(y.shape, self.base_pred)

        for _ in range(self.n_estimators):
            residuals = y - y_pred
            tree = self._split(X, residuals, 0)
            self.trees.append(tree)
            y_pred += self.learning_rate * np.array([self._predict_tree(tree, x) for x in X])

    def predict(self, X):
        y_pred = np.full(X.shape[0], self.base_pred)
        for tree in self.trees:
            y_pred += self.learning_rate * np.array([self._predict_tree(tree, x) for x in X])
        return y_pred


# ------------------------------
# STEP 5: Train and Evaluate
# ------------------------------
model = SimpleXGBoostRegressor(n_estimators=20, learning_rate=0.1, max_depth=3)
model.fit(X, y)

y_pred = model.predict(X)
mse = np.mean((y - y_pred) ** 2)
print(f"✅ Mean Squared Error: {mse:.2f}")

# ------------------------------
# STEP 6: Predict Net Revenue for a Company (User Input)
# ------------------------------
if 'Company' in df.columns:
    company_name = input("Enter the company name: ")

    if company_name not in df['Company'].values:
        print("❌ Company not found in the dataset.")
    else:
        # Select company row and extract features
        company_data = df[df['Company'] == company_name][features].values
        predicted_revenue = model.predict(company_data)[0]

        print(f"\n💼 Company: {company_name}")
        print(f"💰 Predicted Net Revenue: {predicted_revenue:.2f}")

        # Compare with actual (if available)
        actual = df.loc[df['Company'] == company_name, 'Net Revenue'].values[0]
        print(f"📊 Actual Net Revenue: {actual:.2f}")
else:
    print("⚠️ 'Company' column not found in dataset. Please add it to use this feature.")
```

```python
# Step 3: compute Net Revenue
month_cols = ['month_1_revenue','month_2_revenue','month_3_revenue',
              'month_4_revenue','month_5_revenue','month_6_revenue']

# ensure month cols exist
missing_months = [c for c in month_cols if c not in df.columns]
if missing_months:
    raise ValueError("Missing month columns: " + ", ".join(missing_months))

# create Net Revenue
df['Net Revenue'] = df[month_cols].sum(axis=1) - df['Discounts'] - df['Returns & Allowances']

# quick check
print("Net Revenue created. Example values:")
display(df[['Net Revenue']].head())
```

```python
# Step 4: features and dropping missing rows
features = [
    'Operating Expense', 'COGS', 'Discounts', 'Units Sold',
    'MAU/AAU', 'Churn Rate', 'Expansion Revenue'
] + month_cols

# check existence
missing = [c for c in features + ['Net Revenue'] if c not in df.columns]
if missing:
    raise ValueError("Required columns missing: " + ", ".join(missing))

# drop rows where any required field is missing (safer: you can impute instead)
df_clean = df[features + ['Net Revenue']].dropna().reset_index(drop=True)
print("After dropping rows with missing values:", df_clean.shape)
```

```python
# Step 5: prepare arrays
X = df_clean[features].values.astype(float)
y = df_clean['Net Revenue'].values.astype(float)

# sanity
print("X shape:", X.shape, "y shape:", y.shape)
```

```python
from simple_xgb_regressor import SimpleXGBoostRegressor
```

```python
# Step 7: train
model = SimpleXGBoostRegressor(n_estimators=100, learning_rate=0.1, max_depth=3)
model.fit(X, y)

# predictions on training set
y_pred_train = model.predict(X)
mse_train = np.mean((y - y_pred_train)**2)
print("\nFinal training MSE:", mse_train)
```

```python
# Step 8: predict for a company from original df (not df_clean)
# Accept 'Company' or 'name' as the company column
company_col = None
for candidate in ['Company','company','name','Name']:
    if candidate in df.columns:
        company_col = candidate
        break

if company_col is None:
    raise ValueError("No company identifier column found. Add a column named 'Company' or 'name'.")

company_name = input("Enter company name (exact match): ").strip()

# find row(s) in original df
matches = df[df[company_col].astype(str) == company_name]
if matches.shape[0] == 0:
    print("Company not found. Try printing df['{}'].unique() to check names.".format(company_col))
else:
    # pick the first matching row that has the required features non-null
    row = matches.iloc[0]
    # check required features present & not null for this company
    missing_vals = [c for c in features if pd.isna(row.get(c, np.nan))]
    if missing_vals:
        print("Required feature(s) missing for this company:", missing_vals)
    else:
        company_X = np.array([row[features].astype(float).values])
        pred = model.predict(company_X)[0]
        actual = row.get('Net Revenue', np.nan)
        print(f"\nCompany: {company_name}")
        print(f"Predicted Net Revenue: {pred:,.2f}")
        if not pd.isna(actual):
            print(f"Actual Net Revenue (computed): {actual:,.2f}")
```
